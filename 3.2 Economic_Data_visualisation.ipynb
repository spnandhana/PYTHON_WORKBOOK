### Visualization using Matplotlib library

```python
# importing library
import matplotlib.pyplot as plt
```

```python
#ploting of vectors
import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(15,6))
year = np.array([2019,2020,2021,2022,2023,2024,2025])
gdp = np.array([52,57,60,67,70,76,81])
unemployment = np.array([10,12,15,14,20,15,13])
plt.plot(year,gdp,marker ='o',label ="gdp")  
plt.plot(year,unemployment,marker ='o',label ="unemployment")
plt.title("Time series plot of GDP",fontdict= {'fontsize':18,
        'fontweight':'bold',
        'color':'red',
        'verticalalignment': 'baseline',
        'horizontalalignment':'center'})       
plt.ylabel("Gross Domestic product")
plt.xlabel("Year")
plt.legend()
plt.show()
```

### Bar Graph

```python
import matplotlib.pyplot as plt

states = ["Kerala", "Tamil Nadu", "Karnataka", "Maharashtra", "Gujarat"]
gdp = [100,150,180,250,200]

bar = plt.barh(states,gdp)
plt.xlabel("State")
plt.ylabel("GDP")
plt.title("GDP of selected States",fontdict= {'fontsize':19,
        'fontweight':'bold',
        'color':'green',
        'verticalalignment':'baseline',
        'horizontalalignment':'center'})
plt.bar_label(bar)
plt.show()
```

### Scatter Plot

```python
import numpy as np
income = np.array([
    20,20,20,20,
    30,30,30,30,
    40,40,40,40,40,
    50,50,50,50,50,
    60,60,60,60,
    70,70,70,70,70
])

consumption = np.array([
    15,16,13,12,
    18,19,20,25,
    22,23,25,27,26,
    29,30,32,33,35,
    35,36,38,40,
    42,43,45,47,49
])

plt.scatter(income,consumption)

plt.xlabel("income")
plt.xlabel("consumption")
plt.title("income and consumption")

plt.show()
```

### Simple Pie Chart

```python
import matplotlib.pyplot as plt
sectors = ["Education", "Health", "Defence", "Infrastructure"]
expenditure = [30,20,25,25]
plt.pie(expenditure, labels=sectors,autopct="%1.2f%%",explode = [0.1,0,0,0],shadow=True)
plt.title("Government Expenditure")
plt.show()
```

### Visualisation Using Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
data = { "year": [2021,2022,2023,2024,2025],
"GDP": [220,240,260,285,310]}
df = pd.DataFrame(data)
print(df)
```

```python
df.plot(x="year",y="GDP",kind="line",)
plt.title("GDP Trend")
plt.xticks(rotation=20,)
plt.savefig("GDPTrend.png")
plt.show()
```
