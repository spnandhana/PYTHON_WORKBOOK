### 1. String 

- string is a data structure used in python which is enclosed in invered commas 
- for example : "Tom is a good boy" is a string enclosed in double invered commasm

```python
# string Defining
str1 = "Tom is a good boy"
str2 = "India becomes the world's most populus country"
```

#### String operation

##### 1. String Indexing and Slicing

```python
# String Indexing
# 1 positive indexing - starts from left to right from 0,1,2, etc
# 2. negative indexing - starts from right to left of a string from -1 , -2, -3 etc.
print(str1)
print(str1[1])
print(str1[2])
print(str1[3])
```

##### 2. Slicing

```python
print(str1[:3])
print(str1[0:3])
```

```python
print(str1[-3:])
print(str2[-7:])
```
