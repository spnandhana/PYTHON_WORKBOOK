### 03 Tuple

```python
Tuples is an orderd collection of items
```

```python
# creating tuples
num_tup= (1,2,3,4,5,6)
print(num_tup)
# Empty_tuple = ()
# Empty Tuples
empty_tuple = ()
print(empty_tuple)
# Tuple with integers
# Tuple with Strings
str_tup = ("Anandhu", "Bhavya", "Nandana", "Samitha")
print(str_tup)
# mixed tuples
mix_tup = ("Sunil", "Anand", 1,2,5, True, False)
print(mix_tup)
```

```python
# Single element tuple
# One element with comma & Without comma
print(type((10,)))
sing_tup = (10,)
print(sing_tup)
```

```python
# Accesing Elements (indexing)
print(num_tup[0])
print(str_tup[1])
```

```python
#Negative indexing
print(num_tup[-1])
print(num_tup[-2])
```

```python
# slicing
print(num_tup[2:4])
print(str_tup[2:4])
```

```python
# Immutability of Tuple
list1 = list(str_tup) # converting tuple into list
print(list1) # printing list
list1[1] = "Bhavana" # Replacing 2nd item in list1 to "Bhavana"
print(list1)
print(str_tup) # printing str_tup
str_tup[1] = "Bhavana" #Try to replace 2nd element in tuple to "Bhavana"- will show error
```

```python
# Tuple length
print("length of str_tup :",len(str_tup))
print("length of num_tup :",len(num_tup))
```

```python
# Membership checking
print("anu" in str_tup) # from - member in group list ,set,Tuple)
print(1 in num_tup)
```

```python
# Tuple concatenation
print(str_tup)
print(num_tup)
print(str_tup + num_tup)
```

```python
# Tuple Repeating
print(str_tup*3)
print(num_tup*3)
```

```python
# Nested Tuple
nested_tup = ((1,1),(2,3),(3,4))
print(nested_tup)
nested_tup1 = (("Nandhana", "Abhirami"),("Anu","Arun"),((1,1),(2,2),(3,3)),(1,))
print(nested_tup1)
```

```python
# Packing & unpacking
# packing - it is a process of uniting (combinig)distict item in to a tuple data structure
packed = "jose", "Anand", "Nandu", "siva", "poorna", 1,2,3
print(packed)
# unpacking - it is a process to seperate all items from a tuple
print("unpacking of 2 items from a two element tuple")
a,b = ("Economics","Politics")
print(a)
print(b)

print("Unpacking of multiple Items of 8 elements")
a,b,c,d,e,f,g,h = ('jose', 'Anand', 'Nandu', 'siva', 'poorna',1,2,3)
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
print(h)
```

```python
packed = "Anvi", "Anupama", "Raven"
print(packed)
```

```python
a, b, c = packed
print(a)
print(b)
print(c)
```

```python
# Swoping using Tuple
a = (1,2)
print("a = ",a)
b = ("anu", "Maya", "Veena")
print("b = ",b)
a = b
print("After swapping Result")
print("a = ",a)
# length function
print(mix_tup)
print("length of mix_tup:", len(mix_tup))
```

```python
# max function
# max function only work with numeric tuples & str tuple not with mixed tuples
print(num_tup)
print(max(num_tup))
print(str_tup)
print(max(str_tup))
print(mix_tup)
print(max(mix_tup))
```

```python
# min function
print(min(str_tup))
print(min(num_tup))
print(min(mix_tup)) # min function does not work with mixed tuples
```

```python
# Sum function
print(sum(num_tup))
print(sum(str_tup)) # sum function does not work with string tuple and mixed tuple
print(sum(mix_tup))
```

```python
# sorted function
print(sorted(str_tup))
print(sorted(num_tup))
#sorted(mix_tup) # sorted function does not work with mixed tuples
```
