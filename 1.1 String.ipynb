### List Methods

```python
list5 = [1,2,3,4,5,6,20,30, "Aravind", "Nandhana", "Keerthana",]
# Adding a single item - append(x) - append() adds only a single elements to the existing list
list5.append(100)
# Adding multiple items in a list - extend([multiple elements]),
list5.extend(["Aravind", "Niranjana", "Sandhya", 1000,2000])
print(list5)
```

```python
# Adding an item in a pre-assigned position - insert(i,x),
list5.insert(2,"midun")
print(list5)
```

```python
# Removing item - remove(x)
list5.remove("Sandhya")
print(list5)
```

```python
# pop out the i-th indexed items from the list - pop(i)
list5.pop()
print(list5)
# pop out the i-th indexed item from the list pop(i)
list5.pop(-2)
print(list5)
# To get the index of the item in the list x - index(x)
list5.index("Index No.of Niranjana:",list5.index("Niranjana"))
```
