### Demand and Supply Curve-Cost curves-other curves

#### Demand Curve

-The demand curve shows the relationship between price(p)and Quantity Demand(Qd).Accoding to the law of demand other things remainig contact with price
-increases,quantity demand generally decreases.Therefor ,the demand curve generally slopes downward from right to left

```python
import numpy as np
import pandas as pd
p = np.linspace(0,50,100)
Qd = 100 - 2*p
df = pd.DataFrame(zip(Qd,p))
df.columns = ["Quantity_Demanded","Price"]
print(df.head(10))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Price values
p = np.linspace(0,50,100)
# Demand function Qd = f(p)
Qd = 100 - 2 * p

# Plot demand curve
plt.plot(Qd,p, label="Demand Curve")

# Labels
plt.xlabel("Quantity Demanded")
plt.ylabel("Price")

# Title
plt.title("Demand Curve")

# Legend
plt.legend()

# Grid
plt.grid()

# Display graph
plt.show()
```
