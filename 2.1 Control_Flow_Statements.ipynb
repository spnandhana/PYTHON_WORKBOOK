```python
# range function
# range(statrting,ending,interval)
```

```python
# for loop structure
# for iteration_variable_name in data_structure:
for item in range(13,0,-2):
    print(item)
```

```python
# multiplicatin table of 10
num = 10
for item in range(1,11):
    print(num,"X", item, "=", num*item)
```

```python
num = 56
for item in range(1,11):
    print(num,"X", item, "=", num*item)
```

```python
# Additon table
num = 100
for i in range(0,51,5):
    print(num, "+", i, "=", num+i)
```

```python
# Factorial using for loop
# n factorial = n(n-2).....2.1
n=6
fact= 1 #1*2*3
for i in range(1,n+1):
    fact*=i
print(fact)
```

```python
# Factorial using while loops
n=6
fact=1
i=1
while i<=n:
    fact*=i
    i+=1
print(fact)
```

```python
# Continue
dept ="ECONOMICS"
for i in dept:
    if i == "M":
        continue
    print(i)
```

```python
dept = "ECONOMICS"
i=0
while i<len(dept):
    if dept[i]=="M":
        pass
    print(dept[i])
    i+=1
```

```python
dept = "ECONOMICS"
for i in dept:
    if i == "M":
        break
print(i)
```
