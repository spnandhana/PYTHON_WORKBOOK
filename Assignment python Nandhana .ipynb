A) Define p as a symbolic variable using SymPy and specify the demand and supply functions.

```python
import sympy as sp

P = sp.symbols('P')

Qd = 100 - 2*P
Qs = -20 + 2*P

print("Demand Function:", Qd)
print("Supply Function:", Qs)
```

B) Find the equilibrium price by solving:Qd = Qs.

```python
equilibrium_price = sp.solve(Qd - Qs, P)

print("Equilibrium Price:", equilibrium_price[0])
```

### Question 3

A) Use a for-loop to calculate and display the saving of each household.saving = income -- consumption.

```python
households = ["A", "B", "C", "D", "E"]

income = [20000, 30000, 40000, 50000, 60000]

consumption = [18000, 25000, 32000, 38000, 45000]

for i in range(len(households)):
    saving = income[i] - consumption[i]
    print("Household", households[i], "Saving =", saving)
```

B) Use an if-else statement to classify each household as saver or Dissaver.

```python
for i in range(len(households)):
    saving = income[i] - consumption[i]

    if saving >= 0:
        status = "Saver"
    else:
        status = "Dissaver"

    print("Household", households[i], ":", status)
```

C) Define a user-defined function calculate_saving()to calculate household saving.

```python
def saving(income, consumption):
    return income - consumption

print("Saving of Household A:", saving(20000, 18000))
print("Saving of Household B:", saving(30000, 25000))
print("Saving of Household C:", saving(40000, 32000))
```

D) Create a class Household with suitable attributes and methods to calculate and display income,consumption and saving.Create objects for at least three households.

```python
class Household:

    def __init__(self, name, income, consumption):
        self.name = name
        self.income = income
        self.consumption = consumption

    def saving(self):
        return self.income - self.consumption

    def display(self):
        print("Household:", self.name)
        print("Income:", self.income)
        print("Consumption:", self.consumption)
        print("Saving:", self.saving())
        print()


h1 = Household("A", 20000, 18000)
h2 = Household("B", 30000, 25000)
h3 = Household("C", 40000, 32000)

h1.display()
h2.display()
h3.display()
```

E) Use a while loop to display the details of all househols sequentially.

```python
households = ["A", "B", "C", "D", "E"]

income = [20000, 30000, 40000, 50000, 60000]

consumption = [18000, 25000, 32000, 38000, 45000]

i = 0

while i < len(households):

    saving = income[i] - consumption[i]

    print("Household:", households[i])
    print("Income:", income[i])
    print("Consumption:", consumption[i])
    print("Saving:", saving)
    print()

    i = i + 1
```
