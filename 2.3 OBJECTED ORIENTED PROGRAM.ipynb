```python
class calculator:
    pass
import math
from math import sqrt 
class calculator:
    def addition(self,a,b):
        return a+b
    def substractiona(self,a,b):
        return a-b
    def multiplication(self,a,b):
        return a*b
    def divition(self,a,b):
        return a/b
    def squareroot(self,a): 
        return sqrt(a)
calc=calculator()#object or instance
calc.addition(60,23)
calc.divition(10,4)
calc.multiplication(40,39)
calc.squareroot(40)
```
