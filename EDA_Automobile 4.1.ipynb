```python
# To get percentage of null values in a data set
# if null values above 30% drop that columns
# if null values below 30% impute missing values with mean,median or
(auto_df.isnull().sum()/auto_df.shape[0]) * 100
```

```python
for i in auto_df.columns:
    print(auto_df[i].value_counts())
    print("*" * 40)
```

```python
auto_df=auto_df.replace("?",np.nan)
```

```python
auto_df.isnull().sum()/auto_df.shape[0]*100
```

```python
# dropping all the null value contained rows (axis = 0)
auto_df = auto_df.dropna(axis=0)
print(auto_df)
```

```python
auto_df.isnull().sum()
```

```python
num_columns = []
cat_columns = []
for i in auto_df.columns: 
    if auto_df[i].dtype=="int64":
        num_columns.append(i)
    elif auto_df[i].dtype=="float64":
        num_columns.append(i)
    else:
        cat_columns.append(i)
print("Numeric columns:", num_columns)
print("categorical columns:", cat_columns)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
for i in num_columns:
    plt.boxplot(auto_df[i])
    plt.xlabel(i)
    plt.ylabel("numeric values")
    plt.title(i)
    plt.show()
```

```python
for i in cat_columns:
    plt.bar(auto_df[i],height=10)
    plt.xlabel(i)
    plt.ylabel("numeric values")
    plt.title(i)
    plt.show()
```
