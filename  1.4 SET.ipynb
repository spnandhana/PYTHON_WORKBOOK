### SET

```python
A set is an unordered mutable and unndexed collection of unique element in python

stores only unique values
Duplicate values are automatically removed.
does not support indexing or slicing
Elements can be Added or removed
implement using a hash table, making searching very fast
```

```python
# Create a set
numset1={1,2,3,4,5}
print(numset1)
# Empty set
A={}
print(type(A))
B={}
print(type(B))
```

### charactrestics of sets

```python
Unique elements
Unordered
Mutable
No indexing
```

```python
# Unique elements
set1={1,2,3,45,65,14,45,65,12,1,2}
print(set1)
# Unordered
set2 = {1,2,16,3,4,5,16,72,15,14}
print(set2)
# Mutable
set2.add(100)
print(set2)
# No indexing
set[1]
# len
print(len(set1))
print(len(set2))
# max
print(max(set1))
print(max(set2))
# min
print(min(set1))
print(min(set2))
# sum
print(sum(set1))
print(sum(set2))
# sorted
print(sorted(set1))
print(sorted(set2))
# any
print(any(set1))
print(any(set2))
# all
print(all(set1))
print(all(set2))
# type
print(type(set1))
print(type(set2))
```

```python
### A = {1,2,3,4}
B = {3,4,5,6}
print("set A: ", A)
print("set B: ", B)
#1. Union 
print("Union of A and B")
print(A.union(B))
print(A|B)
# 2. Intersection
print("Intersection of A and B")
print(A.intersection(B))
print(A&B)
#3. Difference
print("Difference of A and B (A-B)")
print(A.difference(B))
print(A-B)
print("Difference of B and A (B-A)")
print(B.difference(A))
print(B-A)
# symmetric difference
print("symmetric Difference of A and B")
print(A.symmetric_difference(B))
print(A^B)
```

### Relationship Methods

```python
# issubset
print(A.issubset(B))
# issuperset
print(B.issuperset(A))
# isdisjoint
print(A.isdisjoint(B))
```
