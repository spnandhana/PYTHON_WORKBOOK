## USER DEFINED FUCNTIONS

- A **user-defined function** is a function created by the programmer to perform a specific task.

    - **(i) parameter** - A variable specified in the function definition.
    - Examble:
         ```python
            def add(a, b): 
            ```
- Here, `a` and `b` are **parameters**.

    -**(ii) Argument** - The actual value passed to a function when it is called.
    - Example:
          ```python
           add(5, 8)
          ```
 - Here, `5` and `8` are **arguments**..

### structure of a user_defined_function
```python
def function_name(parameters):
    statements
    return value
```

````python
# user defined function
def grading(mark):
    if mark >= 100:
        return "A+"
    elif mark >= 90:
        return "A"
    elif mark >= 80:
        return "B+"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 50:
        return "D"
    else:
        return "F"
````

````python
grading(90)
````

````python
grading(55)
````

````python
grading(45)
````

````python
grading(60)
````

### Function Rules

- Rule 1: Function begin with def
- Rule 2: Every function must have a name.
- Rule 3: Function name should be meaningful
- Rule 4: Parenthese () are compulsory
- Rule 5: Colon (:) must be used after the function definition.
- Rule 6: Indentation is mandatory
- Rule 7: A function may or may not return a value
- Rule 8: A function executes only when called
- Rule 9: Parameter names should be meaningful.
- Rule 10: Avoid using python keywords as function names

````python
class calculator:
    pass
````

````python
import math
from math import sqrt
class calculator:
    def addition(self,a,b):
        return a+b
    def substraction(self,a,b):
        return a-b
    def multiplication(self,a,b):
        return a*b
    def division(self,a,b):
        return a/b
    def squareroot(self,a):
        return sqrt(a)
calc = calculator() # object or instance
````

````python
calc.addition(5,8)
````

````python
calc.division(4,3)
````

````python
calc.multiplication(20,33)
````

````python
calc.squareroot(8)
````
