### Dictionary

A dictionary is a built-in python data structure that stores data in the form of key-value pairs.

- keys are unique identifiers
- values are the data associated with those keys.
- Dictionaries are mutable,dynamic,and ordered.
- Dictionaries are optimized using hash tables,making key lookup very fast (average O(1)).

```python
# Creating dictionary in key-value pairs {key:values,key:values ...,}
dict1 = {"name": "jose", "age": 30, "course": "python"}
print(dict1)
```

```python
# updating values in the dictionary
dict1["name"] = "Anandhu"
print(dict1)
# change age to 25 & change course to "SQL"
dict1 = {"name": "anandhu", "age":25, "course": "SQL"}
print(dict1)
```

```python
# changing keys
dict1["full_name"] = dict1["name"]
print(dict1)
dict1.pop("name")
print(dict1)
```

### Charactrestic of Dictionary

1. stores data in key-value pairs
2. Mutable - update values in the dictionary and can add new key-value pairs etc.
3. Ordered - dictionary have a definite order. If we add a new key-value pair,it will positioned at last.
4. keys must be unique
5. key must be immutable

#### Creating Dictionary

```python
# Using curly braces
# Using lists
name = ["Anandhu", "Swathy", "Jeeva"]
age = [28,30,31]
course = ["python", "SQL", "powerBI"]
dict2 = {"name": name, "age": age, "course": course}
print(dict2)
# zip method
keys = ["name", "age","course"]
value = ["Swathy", 38, "powerBI"]
dict3 = dict(zip(keys,value))
print(dict3)
# Using tuples
tup = (("name","Swathy"),("age", 30),("Course", "Economics"))
print(dict(tup))
```

```python
# Accessing Dictionary elements
# 1. Using keys
print(dict1["full_name"])
print(dict1["course"])
print(dict1["age"])
# to get keys and values
# 2. Using get
print(dict1.get("age"))
print(dict1.get("course"))
print(dict1.get("full_name"))
# To get all keys
print("All keys in dict1:", dict1.keys())
# To get all values in the dictionary
print("All values in dict1:", dict1.values())
# To convert all dictionary values into list
print(list(dict1.values()))
# 3. To get all the item in the dictionary
print(dict1.items())
```

```python
# Adding new key-value pairs
dict1["place"] = "koothattukulam"
dict1["Gender"] = "Male"
dict1["Education"] = "Graduate"
print(dict1)
```

```python
# Deleting elements in dictionary
# Using del
print(dict1)
del dict1["Education"]
print(dict1)
# Using pop
dict1.pop("Gender")
print(dict1)
# Using pop(specific-key)\
dict1.pop("place")
print(dict1)
# Using pop item()
dict1.popitem()
print(dict1)
# Using clear
```
